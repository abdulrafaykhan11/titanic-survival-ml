```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df['deck']
```

```python
df.drop(['deck','embark_town','alive','adult_male','class','who'],axis=1,inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df.info()
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df['sex'] = le.fit_transform(df['sex'])
df['embarked'] = le.fit_transform(df['embarked']) #s=2,c=0,Q=1
```

```python
df.head()
```

```python
df = df.astype(int)
```

```python
df.head()
```

```python
X = df.drop('survived',axis=1)
y = df['survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
y_train
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled,y_train)
```

```python
y_pred_knn = knn_model.predict(X_test_scaled)
```

```python
print(classification_report(y_test,y_pred_knn))
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model_nb = GaussianNB()
```

```python
model_nb.fit(X_train,y_train)
```

```python
y_nb_pred = model_nb.predict(X_test)
```

```python
y_nb_pred
```

```python
accuracy_score(y_test,y_nb_pred)
```

```python
print(classification_report(y_test,y_nb_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_tree = DecisionTreeClassifier(random_state=42)
```

```python
model_tree.fit(X_train_scaled,y_train)
```

```python
y_pred_tree=  model_tree.predict(X_test_scaled)
```

```python
y_pred_tree
```

```python
accuracy_score(y_test,y_pred_tree)
```

```python
print(classification_report(y_test,y_pred_tree))
```

### Visualizing the Decision Tree

To understand the splits made by the decision tree, we can visualize it. This will show you which features are used and at what thresholds the decisions are made, which are derived from the internal calculations of entropy/information gain.

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20,15))
plot_tree(model_tree, filled=True, feature_names=X.columns.tolist(), class_names=['Not Survived', 'Survived'], rounded=True);
plt.title('Decision Tree Visualization')
plt.show()
```
